```python
# ============================================================
# FILTER DATASET: KEEP ONLY DATE, pH, AND 6 KEY PREDICTORS
# ============================================================

# Define the columns you want to keep:
# - 'Date' (retained for time-series forecasting)
# - 'pH' (target variable)
# - 6 key physical/chemical parameters for prediction
columns_to_keep = [
    "Date",
    "pH",
    "Dissolved_Oxygen_mg_L",
    "Conductivity_uS_cm",
    "Turbidity_NTU",
    "COD_mg_L",
    "Water_Quality_Index",
    "Water_Temperature_C"
]

# Filter the dataframe to only keep columns that exist in your dataset
existing_columns = [col for col in columns_to_keep if col in df.columns]
df_filtered = df[existing_columns].copy()

# Drop rows where essential target data or Date is missing
df_filtered = df_filtered.dropna(subset=["Date", "pH"])

# Verify the new filtered shape and columns
print(f"Original dataset shape: {df.shape}")
print(f"Filtered dataset shape: {df_filtered.shape}")
print("\nRemaining columns:")
print(df_filtered.columns.tolist())

# Preview the filtered data
display(df_filtered.head())
```

# Machine learning 

```python
# Create the binary target variable based on the project definition: pH < 6.5 is Unsafe (1), otherwise Safe (0)
df_filtered["Unsafe_Low_pH"] = (df_filtered["pH"] < 6.5).astype(int)

# Define your 6 selected features + target
features = [
    "Dissolved_Oxygen_mg_L",
    "Conductivity_uS_cm",
    "Turbidity_NTU",
    "COD_mg_L",
    "Water_Quality_Index",
    "Water_Temperature_C"
]

# Ensure only available features are used
features = [f for f in features if f in df_filtered.columns]

X = df_filtered[features].copy()
y = df_filtered["Unsafe_Low_pH"]

# Handle any unexpected missing values by filling with median
X = X.fillna(X.median())

print(f"Feature matrix shape (X): {X.shape}")
print(f"Target vector shape (y): {y.shape}")
print(f"Class distribution:\n{y.value_counts()}")
```

```python
from sklearn.model_selection import train_test_split

# Split data with stratification to maintain class balance
X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y
)

print(f"Training samples: {len(X_train)}")
print(f"Testing samples: {len(X_test)}")
```

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize the Random Forest Classifier
rf_model = RandomForestClassifier(
    n_estimators=100, 
    random_state=42
)

# Train the model on the training data
rf_model.fit(X_train, y_train)

print("Random Forest model successfully trained!")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Make predictions on test data
y_pred = rf_model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%\n")

# Detailed Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=["Safe pH (0)", "Unsafe Low pH (1)"]))

# Plot Confusion Matrix
fig, ax = plt.subplots(figsize=(6, 5))
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Safe", "Unsafe"])
disp.plot(ax=ax, cmap="Blues", colorbar=False)
plt.title("Random Forest Confusion Matrix", fontsize=14, fontweight="bold")
plt.show()
```

```python
# Create a dataframe of feature importances
importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": rf_model.feature_importances_
}).sort_values("Importance", ascending=False)

# Plot feature importances
plt.figure(figsize=(10, 5))
plt.barh(importance_df["Feature"], importance_df["Importance"], color="teal")
plt.xlabel("Importance Score", fontsize=12)
plt.ylabel("Water Parameters", fontsize=12)
plt.title("Feature Importance for pH Prediction", fontsize=14, fontweight="bold")
plt.gca().invert_yaxis()  # Highest importance at the top
plt.grid(axis="x", alpha=0.5)
plt.show()

display(importance_df)
```

# Model evaluation matrix

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Generate predictions and probabilities on the test set
y_pred = rf_model.predict(X_test)
y_proba = rf_model.predict_proba(X_test)[:, 1] # Probability of being Unsafe (class 1)

# Calculate core metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
roc_auc = roc_auc_score(y_test, y_proba)

# Display a formatted metrics summary table
metrics_summary = pd.DataFrame({
    "Evaluation Metric": [
        "Accuracy", 
        "Precision (Unsafe)", 
        "Recall (Unsafe)", 
        "F1-Score (Unsafe)", 
        "ROC-AUC Score"
    ],
    "Score": [
        f"{accuracy:.4f}", 
        f"{precision:.4f}", 
        f"{recall:.4f}", 
        f"{f1:.4f}", 
        f"{roc_auc:.4f}"
    ]
})

print("=== Comprehensive Model Evaluation Summary ===")
display(metrics_summary)
```

```python
# Extract confusion matrix breakdown
cm = confusion_matrix(y_test, y_pred)
tn, fp, fn, tp = cm.ravel()

breakdown_df = pd.DataFrame({
    "Metric Description": [
        "True Negatives (Correctly predicted Safe water)",
        "False Positives (Safe water incorrectly flagged as Unsafe)",
        "False Negatives (Unsafe water missed / flagged as Safe)",
        "True Positives (Correctly predicted Unsafe water)"
    ],
    "Count": [tn, fp, fn, tp]
})

print("=== Confusion Matrix Breakdown ===")
display(breakdown_df)
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

# Compute False Positive Rate and True Positive Rate
fpr, tpr, thresholds = roc_curve(y_test, y_proba)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color="teal", lw=2, label=f"Random Forest (AUC = {roc_auc:.3f})")
plt.plot([0, 1], [0, 1], color="gray", linestyle="--", lw=1, label="Random Guessing (AUC = 0.500)")

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("False Positive Rate (1 - Specificity)", fontsize=12)
plt.ylabel("True Positive Rate (Sensitivity / Recall)", fontsize=12)
plt.title("Receiver Operating Characteristic (ROC) Curve", fontsize=14, fontweight="bold")
plt.legend(loc="lower right", fontsize=12)
plt.grid(alpha=0.5)
plt.show()
```

```python
# ============================================================
# STEP 6: PROPHET TIME-SERIES FORECASTING
# ============================================================

try:
    from prophet import Prophet
    PROPHET_AVAILABLE = True
except ImportError:
    PROPHET_AVAILABLE = False
    print("Prophet library not found. Install it using: pip install prophet")

if PROPHET_AVAILABLE:
    # 1. Prepare time-series data required by Prophet ('ds' for date, 'y' for value)
    ts_data = df[["Date", "pH"]].copy()
    ts_data = ts_data.dropna()
    
    # Aggregate by date if there are multiple records per day
    ts_data = ts_data.groupby("Date")["pH"].mean().reset_index()
    ts_data = ts_data.rename(columns={"Date": "ds", "pH": "y"})
    ts_data = ts_data.sort_values("ds")

    # 2. Initialize and fit the Prophet model
    prophet_model = Prophet()
    prophet_model.fit(ts_data)

    # 3. Create future dataframe for forecasting (e.g., 30 days ahead)
    forecast_days = 30
    future = prophet_model.make_future_dataframe(periods=forecast_days, freq="D")
    forecast = prophet_model.predict(future)

    # 4. Plot the forecast results
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=(12, 6))
    
    historical = forecast[forecast["ds"] <= ts_data["ds"].max()]
    future_forecast = forecast[forecast["ds"] > ts_data["ds"].max()]

    ax.plot(historical["ds"], historical["yhat"], label="Historical / Fitted", color="teal")
    ax.plot(future_forecast["ds"], future_forecast["yhat"], linestyle="--", label="Forecast", color="orange")
    ax.fill_between(
        future_forecast["ds"],
        future_forecast["yhat_lower"],
        future_forecast["yhat_upper"],
        color="orange",
        alpha=0.2,
        label="Uncertainty Interval"
    )
    ax.axhline(6.5, color="red", linestyle=":", label="Unsafe Threshold (pH = 6.5)")

    ax.set_xlabel("Date", fontsize=12)
    ax.set_ylabel("pH", fontsize=12)
    ax.setTitle(f"Water pH {forecast_days}-Day Forecast", fontsize=14, fontweight="bold")
    ax.legend(loc="best")
    ax.grid(True, alpha=0.5)
    plt.show()

    # 5. Display forecast table summary
    print("=== Future pH Forecast Summary (Head) ===")
    display(future_forecast[["ds", "yhat", "yhat_lower", "yhat_upper"]].tail(forecast_days))
```
